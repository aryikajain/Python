# STEMMING:

**what is stemming?**

stemming is a text preprocessing technique in NLP that reduce words to thier root or base form remivung perfizes and suffixes.


**example:**
running
runner
------->
run

**PORTER'S Stemmer**

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

words=["running", "jumps","happily"]

for word in words:
  print(ps.stem(word))

```

**Advantages**

1)Very fast and efficient.

2)Commonly used for tasks like information retrieval and text mining.

**Limitations**

1)Outputs may not always be real words.

2)Limited to English words

**Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer(language='english')
words_to_stem =['running','jumped','happily','quickly','foxes']

for word in words_to_stem:
  print(stemmer.stem(word))
```

**Advantages**

More efficient than Porter Stemmer.

Supports multiple languages.

**Limitations**

More aggressive which might lead to over-stemming.

**Lancaster Stemmer**

```python
from nltk.stem import LancasterStemmer

stemmer = LancasterStemmer()
words_to_stem =['running','jumped','happily','quickly','foxes']

for word in words_to_stem:
  print(stemmer.stem(word))
```

**Advantages**

Very fast.

Good for smaller datasets or quick preprocessing.

**Limitations**

Aggressive which can result in over-stemming.

Less efficient than Snowball in larger datasets.

**Regexp Stemmer**

```python
from nltk.stem import RegexpStemmer

custom_rule = r'ing$'
regexp_stemmer = RegexpStemmer(custom_rule)

word = 'running'
stemmed_word = regexp_stemmer.stem(word)

print(f'Original Word: {word}')
print(f'Stemmed Word: {stemmed_word}')
```

**Advantages**

More accurate, as it preserves linguistic meaning.

Works well with both singular/plural and past/present tense conversions.

**Limitations**

May be inefficient with large corpora.

Slower compared to other stemmers.
